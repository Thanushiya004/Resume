```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_excel("/content/Book1dataset.xlsx")
df
```

```python
df_model = df.drop(columns=["Alert_Sent", "Alert_Type"], errors="ignore")
df_model
```

```python
df_model=df.drop(columns=["Transaction_ID","Customer_ID"],errors="ignore")
df_model
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lab=LabelEncoder()
df['Device_Type']=lab.fit_transform(df['Device_Type'])
df['Country']=lab.fit_transform(df['Country'])
df['Merchant']=lab.fit_transform(df['Merchant'])
df['Merchant_Category']=lab.fit_transform(df['Merchant_Category'])
df
```

```python
if "Transaction_Date" in df.columns:
    df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"], errors="coerce")

    df["Transaction_Hour"] = df["Transaction_Date"].dt.hour
    df["Transaction_Day"] = df["Transaction_Date"].dt.day
    df["Transaction_Month"] = df["Transaction_Date"].dt.month

    df.drop(columns=["Transaction_Date"], inplace=True)
    print(df.head())
```

```python
if "Transaction_Time" in df.columns:
    df["Transaction_Time"] = pd.to_datetime(df["Transaction_Time"], errors="coerce")

    df["Time_hour"] = df["Transaction_Time"].dt.hour
    df["Time_minute"] = df["Transaction_Time"].dt.minute
    df["Time_second"] = df["Transaction_Time"].dt.second

    df.drop(columns=["Transaction_Time"], inplace=True)
    print(df.head())
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(columns=[
    "Is_Fraud",
    "Alert_Sent",
    "Alert_Type",
    "Transaction_ID",
    "Customer_ID"
], errors="ignore")

y = df["Is_Fraud"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
rf=RandomForestClassifier(n_estimators=100,random_state=42)
rf.fit(x_train,y_train)
```

```python
y_pred=rf.predict(x_test)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
acc=accuracy_score(y_test,y_pred)
print("Accuracy:",acc)
```

```python
cm = confusion_matrix(y_test,y_pred)
print("Confusion Matrix:", cm)
```

```python
class_report=classification_report(y_test,y_pred)
print("classification_report",class_report)
```

```python
output = x_test.copy()

output["Actual_Fraud"] = y_test.values
output["Predicted_Fraud"] = y_pred

output["Alert_Sent"] = output["Predicted_Fraud"].apply(
    lambda x: "Yes" if x == 1 else "No"
)

output["Alert_Type"] = output["Predicted_Fraud"].apply(
    lambda x: "SMS/Email" if x == 1 else "None"
)
```

```python
print("\nFraud Detection Alerts:")
print(output.head(10))
```

```python
%%writefile fraud_app.py

import streamlit as st
import numpy as np
import pickle
import pandas as pd
import random
import datetime
import plotly.graph_objects as go
import smtplib
from email.mime.text import MIMEText
from twilio.rest import Client
import random

# ---------------- PAGE CONFIG ----------------
st.set_page_config(
    page_title="Fraud Monitoring Dashboard",
    page_icon="💳",
    layout="wide"
)

# ---------------- DARK THEME + FORCE REMOVE +/- ----------------
st.markdown("""
<style>

/* Dark theme */
body {
    background-color: #0e1117;
    color: white;
}
.stMetric {
    background-color: #1f2937;
    padding: 10px;
    border-radius: 10px;
}

/* FORCE remove number input arrows */
div[data-baseweb="input"] input[type="number"]::-webkit-inner-spin-button,
div[data-baseweb="input"] input[type="number"]::-webkit-outer-spin-button {
    -webkit-appearance: none !important;
    margin: 0 !important;
}

div[data-baseweb="input"] input[type="number"] {
    -moz-appearance: textfield !important;
}

</style>
""", unsafe_allow_html=True)



# ---------------- LOAD MODEL ----------------
model = pickle.load(open("Random_Forest_model.pkl", "rb"))

# ---------------- SESSION STORAGE ----------------
if "history" not in st.session_state:
    st.session_state.history = []

# ---------------- SIDEBAR ----------------
st.sidebar.title("⚙ Dashboard Controls")

mode = st.sidebar.selectbox(
    "Mode",
    ["Live Prediction", "Demo Simulation"]
)

st.sidebar.markdown("---")
st.sidebar.info("Fraud Monitoring System")

# ---------------- HEADER ----------------
st.title("💳 AI Fraud Monitoring Dashboard")

col1, col2 = st.columns([1, 1])

# ================= INPUT PANEL =================
# ================= INPUT PANEL =================
with col1:

    st.subheader("📝 Transaction Input")

    amount = st.text_input("Transaction Amount", "0.00")
    time_gap = st.text_input("Time Since Last Transaction", "14")

    # Safe numeric conversion
    try:
        amount = float(amount)
    except:
        amount = 0.0

    try:
        time_gap = int(time_gap)
    except:
        time_gap = 0

    location = st.selectbox("Location Match", ["Yes", "No"])
    device = st.selectbox("Known Device", ["Yes", "No"])

    location_val = 1 if location == "Yes" else 0
    device_val = 1 if device == "Yes" else 0

    features = np.array([[amount, time_gap, location_val, device_val,
                          0,0,0,0,0,0,0,0]])

    predict = st.button("🔍 Predict Fraud", use_container_width=True)



with col2:

    st.subheader("📊 Risk Visualization")

    if predict:

        result = model.predict(features)

        fraud_prob = random.uniform(0.7, 1.0) if result[0] == 1 else random.uniform(0.0, 0.3)

        # -------- Gauge Meter --------
        fig = go.Figure(go.Indicator(
            mode="gauge+number",
            value=fraud_prob * 100,
            title={'text': "Fraud Risk %"},
            gauge={
                'axis': {'range': [0, 100]},
                'bar': {'color': "red"},
                'steps': [
                    {'range': [0, 40], 'color': "green"},
                    {'range': [40, 70], 'color': "orange"},
                    {'range': [70, 100], 'color': "red"}
                ],
            }
        ))

        st.plotly_chart(fig, use_container_width=True)

        time_now = datetime.datetime.now().strftime("%H:%M:%S")

        # -------- Alert Display --------
        if result[0] == 1:

            st.error("🚨 FRAUD ALERT — Suspicious Activity!")

            st.warning(f"""
            ⚠ Fraud detected
            Time: {time_now}
            Amount: ₹{amount}
            """)

            email_status = "Sent ✅"
            sms_status = "Delivered ✅"

        else:

            st.success("✅ Safe Transaction")

            email_status = "-"
            sms_status = "-"

        # -------- Status Panel --------
        st.info(f"""
        📡 Alert Status
        Email: {email_status}
        SMS: {sms_status}
        """)

        # -------- Save History --------
        st.session_state.history.append({
            "Time": time_now,
            "Amount": amount,
            "Location": location,
            "Device": device,
            "Result": "Fraud" if result[0] == 1 else "Safe"
        })

# ================= ALERT HISTORY =================
st.markdown("---")
st.subheader("📜 Alert History")

if st.session_state.history:
    hist_df = pd.DataFrame(st.session_state.history)
    st.dataframe(hist_df, use_container_width=True)

# ================= DATASET VISUALIZATION =================
st.markdown("---")
st.subheader("📊 Dataset Visualization")

demo_data = pd.DataFrame({
    "Transactions": range(1, 21),
    "Fraud Risk": np.random.rand(20)
})

st.line_chart(demo_data.set_index("Transactions"))

# ================= DEMO SIMULATION =================
if mode == "Demo Simulation":

    st.markdown("---")
    st.subheader("🎯 Live Fraud Simulation")

    sim_data = pd.DataFrame({
        "Transaction": range(1, 11),
        "Risk": np.random.rand(10)
    })

    st.bar_chart(sim_data.set_index("Transaction"))

# ---------------- FOOTER ----------------
st.markdown("---")





```
